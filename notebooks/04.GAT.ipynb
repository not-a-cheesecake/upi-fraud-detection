```python

import numpy as np
import torch

num_nodes = num_accounts

context_src = source_ids[context_positions]
context_dst = destination_ids[context_positions]
context_amount = df["amount"].to_numpy(dtype=np.float32)[context_positions]

sent_count = np.bincount(
    context_src, minlength=num_nodes
).astype(np.float32)

received_count = np.bincount(
    context_dst, minlength=num_nodes
).astype(np.float32)

sent_amount = np.bincount(
    context_src,
    weights=context_amount,
    minlength=num_nodes
).astype(np.float32)

received_amount = np.bincount(
    context_dst,
    weights=context_amount,
    minlength=num_nodes
).astype(np.float32)

node_features = torch.tensor(
    np.column_stack([
        sent_count,
        received_count,
        sent_amount,
        received_amount
    ]),
    dtype=torch.float32
)

print("Node feature shape:", tuple(node_features.shape))
print("Data type:", node_features.dtype)
print("All finite:", bool(torch.isfinite(node_features).all()))
print("Nodes with historical activity:", int((node_features.sum(dim=1) > 0).sum()))
```

```python

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import GATConv

class TransactionGAT(nn.Module):
    def __init__(
        self,
        node_features=4,
        transaction_features=10,
        hidden=16,
        heads=2
    ):
        super().__init__()

        self.gat1 = GATConv(
            node_features,
            hidden,
            heads=heads,
            dropout=0.2
        )

        self.gat2 = GATConv(
            hidden * heads,
            hidden,
            heads=1,
            concat=False,
            dropout=0.2
        )

        self.classifier = nn.Sequential(
            nn.Linear(hidden * 2 + transaction_features, hidden),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(hidden, 2)
        )

    def forward(
        self,
        x,
        edge_index,
        source_ids,
        destination_ids,
        transaction_features
    ):
        x = F.elu(self.gat1(x, edge_index))
        x = F.elu(self.gat2(x, edge_index))

        source_embedding = x[source_ids]
        destination_embedding = x[destination_ids]

        combined = torch.cat(
            [
                source_embedding,
                destination_embedding,
                transaction_features
            ],
            dim=1
        )

        return self.classifier(combined)


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

gat_model = TransactionGAT().to(device)

print(gat_model)
print("\nDevice:", next(gat_model.parameters()).device)
print("Trainable parameters:",
      sum(p.numel() for p in gat_model.parameters() if p.requires_grad))
```

```python

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import GATConv

class TransactionGAT(nn.Module):
    def __init__(
        self,
        node_features=4,
        transaction_features=10,
        hidden=16,
        heads=2
    ):
        super().__init__()

        self.gat1 = GATConv(
            node_features, hidden,
            heads=heads, dropout=0.2
        )

        self.gat2 = GATConv(
            hidden * heads, hidden,
            heads=1, concat=False, dropout=0.2
        )

        self.classifier = nn.Sequential(
            nn.Linear(hidden * 2 + transaction_features, hidden),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(hidden, 2)
        )

    def forward(
        self, x, edge_index,
        source_ids, destination_ids,
        transaction_features
    ):
        x = F.elu(self.gat1(x, edge_index))
        x = F.elu(self.gat2(x, edge_index))

        source_embedding = x[source_ids]
        destination_embedding = x[destination_ids]

        combined = torch.cat(
            [
                source_embedding,
                destination_embedding,
                transaction_features
            ],
            dim=1
        )

        return self.classifier(combined)


gat_model = TransactionGAT().to(device)

print("GAT model created successfully.")
print("Device:", next(gat_model.parameters()).device)
print("Trainable parameters:",
      sum(p.numel() for p in gat_model.parameters() if p.requires_grad))
```

```python

class TransactionGAT(nn.Module):
    def __init__(
        self,
        node_features=4,
        transaction_features=10,
        hidden=16,
        heads=2
    ):
        super().__init__()

        self.gat1 = GATConv(
            node_features, hidden,
            heads=heads, dropout=0.2
        )

        self.gat2 = GATConv(
            hidden * heads, hidden,
            heads=1, concat=False, dropout=0.2
        )

        self.classifier = nn.Sequential(
            nn.Linear(hidden * 2 + transaction_features, hidden),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(hidden, 2)
        )

    def forward(
        self, x, edge_index,
        source_ids, destination_ids,
        transaction_features
    ):
        x = F.elu(self.gat1(x, edge_index))
        x = F.elu(self.gat2(x, edge_index))

        source_embedding = x[source_ids]
        destination_embedding = x[destination_ids]

        combined = torch.cat(
            [
                source_embedding,
                destination_embedding,
                transaction_features
            ],
            dim=1
        )

        return self.classifier(combined)


config = checkpoint["model_config"]

gat_model = TransactionGAT(**config)
gat_model.load_state_dict(checkpoint["model_state_dict"])
gat_model = gat_model.to("cuda")
gat_model.eval()

transaction_scaler = checkpoint["transaction_scaler"]

print("GAT model restored successfully!")
print("Device:", next(gat_model.parameters()).device)
print("Model in evaluation mode:", not gat_model.training)
print("Scaler restored:", transaction_scaler is not None)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import GATConv

class TransactionGAT(nn.Module):
    def __init__(self, node_features=4, transaction_features=10, hidden=16, heads=2):
        super().__init__()
        self.gat1 = GATConv(node_features, hidden, heads=heads, dropout=0.2)
        self.gat2 = GATConv(hidden * heads, hidden, heads=1,
                            concat=False, dropout=0.2)
        self.classifier = nn.Sequential(
            nn.Linear(hidden * 2 + transaction_features, hidden),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(hidden, 2)
        )

    def forward(self, x, edge_index, source_ids, destination_ids, transaction_features):
        x = F.elu(self.gat1(x, edge_index))
        x = F.elu(self.gat2(x, edge_index))
        source_embedding = x[source_ids]
        destination_embedding = x[destination_ids]
        combined = torch.cat(
            [source_embedding, destination_embedding, transaction_features],
            dim=1
        )
        return self.classifier(combined)

config = checkpoint["model_config"]
gat_model = TransactionGAT(**config)
gat_model.load_state_dict(checkpoint["model_state_dict"])
gat_model = gat_model.to("cuda")
gat_model.eval()

transaction_scaler = checkpoint["transaction_scaler"]

print("Model restored successfully!")
print("Device:", next(gat_model.parameters()).device)
print("Evaluation mode:", not gat_model.training)
print("Scaler restored:", transaction_scaler is not None)
```

```python

# Ek validation fraud transaction select karna
target_transaction = validation_fraud.iloc[0]

print("Selected transaction:")
print("Step:", target_transaction["step"])
print("Type:", target_transaction["type"])
print("Amount:", target_transaction["amount"])
print("Fraud label:", target_transaction["isFraud"])
print("Source account:", target_transaction["nameOrig"])
print("Destination account:", target_transaction["nameDest"])
```

```python

import numpy as np
import pandas as pd

# All accounts ko consistent numeric IDs dena
all_accounts = pd.concat(
    [fraud_sample["nameOrig"], fraud_sample["nameDest"]],
    ignore_index=True
)

account_codes, unique_accounts = pd.factorize(all_accounts, sort=False)

num_transactions = len(fraud_sample)

source_ids = account_codes[:num_transactions].astype(np.int64)
destination_ids = account_codes[num_transactions:].astype(np.int64)

num_accounts = len(unique_accounts)

# Sirf historical context: steps 1–279
context_mask = fraud_sample["step"].between(1, 279).to_numpy()
context_positions = np.flatnonzero(context_mask)

context_src = source_ids[context_positions]
context_dst = destination_ids[context_positions]

print("Total accounts:", num_accounts)
print("Historical transactions:", len(context_positions))
print("Historical edges:", len(context_src))
print("Account IDs prepared successfully!")

```

```python

# Historical account features only
context_amount = fraud_sample["amount"].to_numpy(
    dtype=np.float32
)[context_positions]

sent_count = np.bincount(
    context_src, minlength=num_accounts
).astype(np.float32)

received_count = np.bincount(
    context_dst, minlength=num_accounts
).astype(np.float32)

sent_amount = np.bincount(
    context_src,
    weights=context_amount,
    minlength=num_accounts
).astype(np.float32)

received_amount = np.bincount(
    context_dst,
    weights=context_amount,
    minlength=num_accounts
).astype(np.float32)

node_features = np.column_stack([
    sent_count,
    received_count,
    sent_amount,
    received_amount
])

print("Node features shape:", node_features.shape)
print("All values finite:", np.isfinite(node_features).all())
print("Active historical accounts:",
      int(np.count_nonzero(sent_count + received_count)))
```

```python

import torch
from torch_geometric.data import Data

# Account features ko PyTorch tensor mein convert karna
node_features_tensor = torch.tensor(
    node_features,
    dtype=torch.float32
)

# Historical transaction edges
context_edge_index = torch.tensor(
    np.stack([context_src, context_dst]),
    dtype=torch.long
)

graph_data = Data(
    x=node_features_tensor,
    edge_index=context_edge_index,
    num_nodes=num_accounts
)

print("Historical graph created successfully!")
print("Graph nodes:", graph_data.num_nodes)
print("Historical edges:", graph_data.edge_index.shape[1])
print("Node feature shape:", tuple(graph_data.x.shape))
print("Edge IDs in valid range:",
      int(graph_data.edge_index.max()) < graph_data.num_nodes)
```

```python

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import GATConv

class TransactionGAT(nn.Module):
    def __init__(
        self,
        node_features=4,
        transaction_features=10,
        hidden=16,
        heads=2
    ):
        super().__init__()

        self.gat1 = GATConv(
            node_features, hidden,
            heads=heads, dropout=0.2
        )
        self.gat2 = GATConv(
            hidden * heads, hidden,
            heads=1, concat=False, dropout=0.2
        )
        self.classifier = nn.Sequential(
            nn.Linear(hidden * 2 + transaction_features, hidden),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(hidden, 2)
        )

    def forward(
        self, x, edge_index, source_ids,
        destination_ids, transaction_features
    ):
        x = F.elu(self.gat1(x, edge_index))
        x = F.elu(self.gat2(x, edge_index))

        source_embedding = x[source_ids]
        destination_embedding = x[destination_ids]

        combined = torch.cat(
            [source_embedding, destination_embedding,
             transaction_features],
            dim=1
        )
        return self.classifier(combined)


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
config = checkpoint["model_config"]

gat_model = TransactionGAT(**config).to(device)
gat_model.load_state_dict(checkpoint["model_state_dict"])
gat_model.eval()

transaction_scaler = checkpoint["transaction_scaler"]
threshold = checkpoint["threshold"]

print("GAT model restored successfully!")
print("Device:", device)
print("Epoch:", checkpoint["epoch"])
print("Threshold:", threshold)
print("Model in evaluation mode:", not gat_model.training)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import GATConv

class TransactionGAT(nn.Module):
    def __init__(
        self,
        node_features=4,
        transaction_features=10,
        hidden=16,
        heads=2
    ):
        super().__init__()

        self.gat1 = GATConv(
            node_features, hidden,
            heads=heads, dropout=0.2
        )

        self.gat2 = GATConv(
            hidden * heads, hidden,
            heads=1, concat=False, dropout=0.2
        )

        self.classifier = nn.Sequential(
            nn.Linear(hidden * 2 + transaction_features, hidden),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(hidden, 2)
        )

    def forward(
        self, x, edge_index,
        source_ids, destination_ids,
        transaction_features
    ):
        x = F.elu(self.gat1(x, edge_index))
        x = F.elu(self.gat2(x, edge_index))

        source_embedding = x[source_ids]
        destination_embedding = x[destination_ids]

        combined = torch.cat(
            [
                source_embedding,
                destination_embedding,
                transaction_features
            ],
            dim=1
        )

        return self.classifier(combined)

print("TransactionGAT class defined successfully.")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch_geometric
from torch_geometric.nn import GATConv

print("PyTorch:", torch.__version__)
print("PyG:", torch_geometric.__version__)
print("CUDA available:", torch.cuda.is_available())

class TransactionGAT(nn.Module):
    def __init__(
        self, node_features=4,
        transaction_features=10,
        hidden=16, heads=2
    ):
        super().__init__()
        self.gat1 = GATConv(
            node_features, hidden,
            heads=heads, dropout=0.2
        )
        self.gat2 = GATConv(
            hidden * heads, hidden,
            heads=1, concat=False, dropout=0.2
        )
        self.classifier = nn.Sequential(
            nn.Linear(hidden * 2 + transaction_features, hidden),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(hidden, 2)
        )

    def forward(
        self, x, edge_index,
        source_ids, destination_ids,
        transaction_features
    ):
        x = F.elu(self.gat1(x, edge_index))
        x = F.elu(self.gat2(x, edge_index))
        source_embedding = x[source_ids]
        destination_embedding = x[destination_ids]
        combined = torch.cat(
            [source_embedding, destination_embedding,
             transaction_features],
            dim=1
        )
        return self.classifier(combined)

print("TransactionGAT class defined successfully.")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import GATConv

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

class TransactionGAT(nn.Module):
    def __init__(self, node_features=4, transaction_features=10,
                 hidden=16, heads=2):
        super().__init__()
        self.gat1 = GATConv(
            node_features, hidden, heads=heads, dropout=0.2
        )
        self.gat2 = GATConv(
            hidden * heads, hidden, heads=1,
            concat=False, dropout=0.2
        )
        self.classifier = nn.Sequential(
            nn.Linear(hidden * 2 + transaction_features, hidden),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(hidden, 2)
        )

    def forward(self, x, edge_index, source_ids,
                destination_ids, transaction_features):
        x = F.elu(self.gat1(x, edge_index))
        x = F.elu(self.gat2(x, edge_index))
        source_embedding = x[source_ids]
        destination_embedding = x[destination_ids]
        combined = torch.cat(
            [source_embedding, destination_embedding,
             transaction_features], dim=1
        )
        return self.classifier(combined)

config = checkpoint["model_config"]

gat_model = TransactionGAT(
    node_features=config["node_features"],
    transaction_features=config["transaction_features"],
    hidden=config["hidden"],
    heads=config["heads"]
).to(device)

gat_model.load_state_dict(checkpoint["model_state_dict"])
gat_model.eval()

transaction_scaler = checkpoint["transaction_scaler"]
threshold = checkpoint["threshold"]

print("Model restored successfully!")
print("Device:", device)
print("Checkpoint epoch:", checkpoint["epoch"])
print("Threshold:", threshold)
print("Validation PR-AUC:", checkpoint["validation_pr_auc"])
print("Validation ROC-AUC:", checkpoint["validation_roc_auc"])
print("Evaluation mode:", not gat_model.training)
```

```python

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ---------- Folders ----------
base = "/content/drive/MyDrive/UPI_Fraud_Project"
source = os.path.join(base, "explainability_deliverables")
out = os.path.join(base, "gat_graphs")
os.makedirs(out, exist_ok=True)

test_path = os.path.join(source, "final_test_results.csv")
threshold_path = os.path.join(source, "gat_validation_threshold_results.csv")

if not os.path.exists(test_path):
    raise FileNotFoundError(f"Missing file: {test_path}")
if not os.path.exists(threshold_path):
    raise FileNotFoundError(f"Missing file: {threshold_path}")

test_df = pd.read_csv(test_path)
threshold_df = pd.read_csv(threshold_path)

print("Test results columns:", test_df.columns.tolist())
print("Threshold results columns:", threshold_df.columns.tolist())

# ---------- 1. Final test performance ----------
# Use actual saved metrics when recognizable; known saved results as fallback.
known_metrics = {
    "PR-AUC": 0.685268,
    "ROC-AUC": 0.812556,
    "Precision": 0.916696,
    "Recall": 0.617661,
    "F1": 0.738038
}

def find_column(df, keywords):
    for col in df.columns:
        name = str(col).strip().lower().replace("_", "-")
        if any(k in name for k in keywords):
            return col
    return None

metric_aliases = {
    "PR-AUC": ["pr-auc", "prauc", "average-precision"],
    "ROC-AUC": ["roc-auc", "rocauc"],
    "Precision": ["precision"],
    "Recall": ["recall"],
    "F1": ["f1"]
}

metrics = {}
for label, aliases in metric_aliases.items():
    col = find_column(test_df, aliases)
    if col is not None:
        try:
            val = pd.to_numeric(test_df[col], errors="coerce").dropna()
            if len(val):
                metrics[label] = float(val.iloc[0])
        except Exception:
            pass

# Fill only metrics not present in the saved result file.
for label, value in known_metrics.items():
    metrics.setdefault(label, value)

plt.figure(figsize=(9, 5))
bars = plt.bar(metrics.keys(), metrics.values())
plt.ylim(0, 1)
plt.ylabel("Score")
plt.title("GAT Model — Final Test Performance")
plt.grid(axis="y", alpha=0.25)
for bar, value in zip(bars, metrics.values()):
    plt.text(bar.get_x() + bar.get_width()/2, value + 0.02,
             f"{value:.3f}", ha="center", fontsize=9)
plt.tight_layout()
plt.savefig(os.path.join(out, "01_final_test_metrics.png"), dpi=200)
plt.close()

# ---------- 2. Confusion matrix ----------
# Counts from the saved epoch-4 final test evaluation.
tn, fp, fn, tp = 1289026, 239, 1628, 2630
cm = np.array([[tn, fp], [fn, tp]])

fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(cm, interpolation="nearest")
ax.set_title("GAT Confusion Matrix — Final Test")
ax.set_xlabel("Predicted label")
ax.set_ylabel("Actual label")
ax.set_xticks([0, 1], labels=["Non-fraud", "Fraud"])
ax.set_yticks([0, 1], labels=["Non-fraud", "Fraud"])

for i in range(2):
    for j in range(2):
        ax.text(j, i, f"{cm[i, j]:,}", ha="center", va="center",
                color="white" if cm[i, j] > cm.max()/2 else "black")
fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
fig.tight_layout()
fig.savefig(os.path.join(out, "02_confusion_matrix.png"), dpi=200)
plt.close(fig)

# ---------- 3. Validation threshold analysis ----------
df = threshold_df.copy()
df.columns = [str(c).strip() for c in df.columns]

threshold_col = find_column(df, ["threshold"])
if threshold_col is None:
    # Try common abbreviated column names.
    threshold_col = next(
        (c for c in df.columns if str(c).strip().lower() in ["thr", "thresh"]),
        None
    )

if threshold_col is not None:
    x = pd.to_numeric(df[threshold_col], errors="coerce")
    metric_cols = []
    for col in df.columns:
        if col == threshold_col:
            continue
        name = str(col).lower().replace("_", "-")
        if any(k in name for k in ["precision", "recall", "f1", "pr-auc", "prauc"]):
            if pd.to_numeric(df[col], errors="coerce").notna().any():
                metric_cols.append(col)

    if metric_cols:
        fig, ax = plt.subplots(figsize=(10, 6))
        for col in metric_cols:
            y = pd.to_numeric(df[col], errors="coerce")
            ax.plot(x, y, marker="o", linewidth=1.5, label=str(col))
        ax.set_xlabel("Decision threshold")
        ax.set_ylabel("Metric value")
        ax.set_title("GAT Validation — Metrics by Threshold")
        ax.grid(alpha=0.3)
        ax.legend()
        fig.tight_layout()
        fig.savefig(os.path.join(out, "03_validation_threshold_analysis.png"),
                    dpi=200)
        plt.close(fig)
        print("Created threshold analysis graph.")
    else:
        print("No recognizable metric columns found for threshold graph.")
else:
    print("No threshold column recognized; see printed column names above.")

# ---------- 4. Save source tables and graph notes ----------
test_df.to_csv(os.path.join(out, "final_test_results_copy.csv"), index=False)
threshold_df.to_csv(
    os.path.join(out, "validation_threshold_results_copy.csv"), index=False
)

readme = """# GAT Result Graphs

This folder contains visualizations of the current GAT evaluation results.

- 01_final_test_metrics.png: saved final test metrics.
- 02_confusion_matrix.png: final test confusion matrix.
- 03_validation_threshold_analysis.png: generated when recognizable threshold
  and metric columns are present.
- CSV copies preserve the source result tables.

Model: epoch 4, decision threshold 0.75.
Final test: PR-AUC 0.685268, ROC-AUC 0.812556,
Precision 0.916696, Recall 0.617661, F1 0.738038.

Note: These are evaluation visualizations, not GNNExplainer feature attributions.
Raw prediction scores are needed to reconstruct genuine ROC and precision-recall
curves. GNNExplainer attribution output is still pending.
"""
with open(os.path.join(out, "GRAPH_README.md"), "w") as f:
    f.write(readme)

print("\nSaved files in:", out)
print(*sorted(os.listdir(out)), sep="\n")
```
