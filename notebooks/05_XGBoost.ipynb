```python

from pathlib import Path
import joblib
import json

project = Path("/content/drive/MyDrive/UPI_Fraud_Project")
out_dir = project / "xgboost_deliverables"
out_dir.mkdir(parents=True, exist_ok=True)

model_path = project / "explainability_deliverables/xgboost_temporal_model.joblib"
prep_path = project / "explainability_deliverables/xgboost_temporal_preprocessor.joblib"

model = joblib.load(model_path)
preprocessor = joblib.load(prep_path)

summary = {
    "model_type": type(model).__name__,
    "model_parameters": model.get_params(),
    "preprocessor_input_features": preprocessor.feature_names_in_.tolist(),
    "transformed_features": preprocessor.get_feature_names_out().tolist(),
    "n_features": int(model.n_features_in_),
}

# Make parameters JSON-safe (for example, missing=np.nan).
summary["model_parameters"] = {
    key: str(value) if value is None or (
        isinstance(value, float) and value != value
    ) else value
    for key, value in summary["model_parameters"].items()
}

with open(out_dir / "xgboost_model_configuration.json", "w") as f:
    json.dump(summary, f, indent=2)

print(json.dumps({
    "model_type": summary["model_type"],
    "n_features": summary["n_features"],
    "preprocessor_input_features": summary["preprocessor_input_features"],
    "transformed_features": summary["transformed_features"],
}, indent=2))

print("\nConfiguration saved to:", out_dir / "xgboost_model_configuration.json")
```

```python

import numpy as np
import pandas as pd
import json

from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
)

DATA_PATH = "/content/drive/MyDrive/PS_20174392719_1491204439457_log.csv"
OUT_DIR = project / "xgboost_deliverables"

# Load only the test period needed for evaluation
df_test = pd.read_csv(DATA_PATH)
df_test = df_test[df_test["step"].between(355, 743)].copy()

features = preprocessor.feature_names_in_.tolist()
X_test_raw = df_test[features]
y_test = df_test["isFraud"].to_numpy()

# Use the existing fitted preprocessor and saved model
X_test = preprocessor.transform(X_test_raw)
test_prob = model.predict_proba(X_test)[:, 1]

# Use the existing project threshold of 0.5 for this evaluation
threshold = 0.5
test_pred = (test_prob >= threshold).astype("int8")

tn, fp, fn, tp = confusion_matrix(
    y_test, test_pred, labels=[0, 1]
).ravel()

results = {
    "model": "XGBoost",
    "split": "Temporal",
    "threshold": threshold,
    "test_transactions": int(len(y_test)),
    "actual_fraud": int(y_test.sum()),
    "precision": float(precision_score(y_test, test_pred, zero_division=0)),
    "recall": float(recall_score(y_test, test_pred, zero_division=0)),
    "f1": float(f1_score(y_test, test_pred, zero_division=0)),
    "pr_auc": float(average_precision_score(y_test, test_prob)),
    "roc_auc": float(roc_auc_score(y_test, test_prob)),
    "true_negative": int(tn),
    "false_positive": int(fp),
    "false_negative": int(fn),
    "true_positive": int(tp),
    "flagged_transactions": int(test_pred.sum()),
}

print(json.dumps(results, indent=2))

OUT_DIR.mkdir(parents=True, exist_ok=True)
pd.DataFrame([results]).to_csv(
    OUT_DIR / "xgboost_temporal_evaluation.csv",
    index=False
)

pd.DataFrame({
    "actual_label": y_test,
    "fraud_probability": test_prob,
    "predicted_label": test_pred,
}).to_csv(OUT_DIR / "xgboost_test_predictions.csv", index=False)

print("\nSaved XGBoost evaluation files to:", OUT_DIR)
```

```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, test_pred, labels=[0, 1])

fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(cm)
ax.set_title("XGBoost — Temporal Test Confusion Matrix")
ax.set_xlabel("Predicted label")
ax.set_ylabel("Actual label")
ax.set_xticks([0, 1], ["Genuine", "Fraud"])
ax.set_yticks([0, 1], ["Genuine", "Fraud"])

for i in range(2):
    for j in range(2):
        ax.text(j, i, f"{cm[i, j]:,}", ha="center", va="center")

fig.colorbar(im, ax=ax)
fig.tight_layout()
fig.savefig(OUT_DIR / "xgboost_confusion_matrix.png", dpi=160)
plt.show()

print("Saved:", OUT_DIR / "xgboost_confusion_matrix.png")
```
